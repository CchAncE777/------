```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# Скачивание данных Natural Earth
url = "https://naturalearth.s3.amazonaws.com/110m_cultural/ne_110m_admin_0_countries.zip"
world = gpd.read_file(url)

# Пример датафрейма с данными
data = {
    "playerid": [1, 2, 3, 4, 5, 6],
    "country": ["United States", "Russia", "Germany", "United States", "Russia", "Japan"]
}

df = pd.DataFrame(data)

# Подсчёт количества пользователей по странам
country_stats = df['country'].value_counts().reset_index()
country_stats.columns = ['country', 'user_count']

# Соединение статистики с геоданными
merged = world.merge(country_stats, left_on='ADMIN', right_on='country', how='left')

# Построение карты
fig, ax = plt.subplots(1, 1, figsize=(15, 10))
merged.plot(column='user_count', ax=ax, cmap='plasma', legend=True)
ax.set_title("Steam Users by Country", fontsize=15)
plt.show()
```
